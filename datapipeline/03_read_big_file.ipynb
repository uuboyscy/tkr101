```python
filter_portno_2998623 = df["portno"] == "2998623"
filter_percent_tna_gt_2 = df["percent_tna"] > 2
filter_rank_gte_2 = df["security_rank"] >= 2
filter_rank_lte_4 = df["security_rank"] <= 4
```

```python
"""
select *
from df
where (security_rank >= 2 and security_rank <= 4)
or (portno = '2998623' and percent_tna > 2)
"""
df[
    (filter_rank_gte_2 & filter_rank_lte_4)
    | (filter_portno_2998623 & filter_percent_tna_gt_2)
]
```

## groupby

```python
df.groupby("portno")
```

```python
"""
select count(1)
from df
group by portno;
"""
df.groupby("portno").size()
```

```python
"""
select
    portno,
    count(1)
from df
group by portno;
"""
df.groupby("portno", as_index=False).size()
```

```python
"""
select
    portno,
    sum()
from df
group by portno;
"""
df.groupby("portno", as_index=False).sum()
```

```python
"""
select
    portno,
    sum(security_rank),
    sum(percent_tna)
from df
group by portno;
"""
df.groupby("portno", as_index=False)[["security_rank", "percent_tna"]].sum()
```

```python
df.groupby(["portno", "eff_dt"])["percent_tna"].sum()
```

```python
df.groupby(["portno", "eff_dt"], as_index=False)["percent_tna"].sum()
```

```python
df.groupby(["portno", "eff_dt"], as_index=False)["percent_tna"].sum().sort_values(["portno", "eff_dt"], ascending=False)
```

```python
df.groupby(["portno", "eff_dt"], as_index=False)["percent_tna"].sum().sort_values(["portno", "eff_dt"], ascending=False)
df_t = (
    df
    .groupby(
        [
            "portno",
            "eff_dt",
        ],
        as_index=False,
    )["percent_tna"]
    .sum()
    .sort_values(
        [
            "portno",
            "eff_dt",
        ],
        ascending=False,
    )
)
```
