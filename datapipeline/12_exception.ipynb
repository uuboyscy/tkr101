```python
try:
    1/0
except ZeroDivisionError as e:
    print(e)
    with open("error.log", "w", encoding="utf-8") as f:
        f.write(f"[ERROR] | yyyy-mm-dd HH:MM:ss | {e} | <{__name__}>")
except Exception as e:
    print(e)
    with open("error.log", "w", encoding="utf-8") as f:
        f.write(f"[ERROR] | yyyy-mm-dd HH:MM:ss | {e} | <{__name__}>")
except KeyboardInterrupt as e:
    print(e)
    with open("error.log", "w", encoding="utf-8") as f:
        f.write(f"[CRITICAL] | yyyy-mm-dd HH:MM:ss | {e} | <{__name__}>")
```
